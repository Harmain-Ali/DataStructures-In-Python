# SINGLY LINKED LIST


## BASIC STRUCTURE FOR CREATING A SINGLY LINKED LIST

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, data):
        """Appends a new node to the end of the doubly linked list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            self.tail = new_node
            return

        self.tail.next = new_node
        new_node.prev = self.tail
        self.tail = new_node

    def display_forward(self):
        """Prints the doubly linked list from head to tail."""
        current = self.head

        if not current:
            print("List is empty!")
            return

        while current:
            print(f"{current.data} <-> ", end="")
            current = current.next
        print("None")

    def display_backward(self):
        """Prints the doubly linked list from tail to head."""
        current = self.tail

        if not current:
            print("List is empty!")
            return

        while current:
            print(f"{current.data} <-> ", end="")
            current = current.prev
        print("None")


# --- Testing the implementation ---
dll = DoublyLinkedList()

# Case 1: Empty list display
dll.display_forward()  # Output: List is empty!

# Case 2: Populated list display
dll.append(10)
dll.append(20)
dll.append(30)

# Display forward from head
dll.display_forward()  # Output: 10 <-> 20 <-> 30 <-> None

# Display backward from tail
dll.display_backward()  # Output: 30 <-> 20 <-> 10 <-> None
````

### Detailed Technical Breakdown

* **`new_node = Node(data)`**
**Dynamic Memory Allocation:** Allocates heap memory for `new_node` with `data`, setting both `new_node.next` and `new_node.prev` to `None`.
* **`if self.head is None:`**
**Base Case / Empty List Guard:** Checks if the data structure contains zero nodes.
* **`self.head = new_node` & `self.tail = new_node**`
If empty, assigns `new_node` to both `head` and `tail` pointers simultaneously, completing $O(1)$ setup.
* **`self.tail.next = new_node`**
**Forward Link Creation:** Updates the current last node's (`tail`) forward reference pointer (`next`) to point to the memory address of `new_node`.
* **`new_node.prev = self.tail`**
**Backward Link Creation:** Sets the new node's backward reference pointer (`prev`) to point back to the existing `tail` node address.
* **`self.tail = new_node`**
**Tail Marker Update:** Updates the list's `tail` tracking attribute to point to `new_node`, making it the new official end of the list.

> **Time Complexity:** $O(1)$ — Because a `tail` pointer is maintained, appending takes constant time and does not require traversing the list.

## INSERTING A NODE AT THE START

````python

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def insert_at_beginning(self, data):
        """Inserts a new node at the start (head) of the doubly linked list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            self.tail = new_node
            return

        new_node.next = self.head
        self.head.prev = new_node
        self.head = new_node

    def display_forward(self):
        """Prints the doubly linked list from head to tail."""
        current = self.head

        if not current:
            print("List is empty!")
            return

        while current:
            print(f"{current.data} <-> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
dll = DoublyLinkedList()

# Case 1: Inserting into an empty list
dll.insert_at_beginning(30)
dll.display_forward()  # Output: 30 <-> None

# Case 2: Inserting additional nodes at the start
dll.insert_at_beginning(20)
dll.insert_at_beginning(10)

# Display final structure
dll.display_forward()  # Output: 10 <-> 20 <-> 30 <-> None
````

## INSERTING A NODE AFTER AN INDEX

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def insert_after_index(self, index, data):
        """Inserts a new node immediately after the target index (0-based)."""
        # Guard Clause 1: Empty list check
        if self.head is None:
            print("List is empty! Cannot insert.")
            return False

        # Guard Clause 2: Negative index check
        if index < 0:
            print("Index cannot be negative!")
            return False

        current = self.head
        current_index = 0

        # Linear traversal to locate target index
        while current and current_index < index:
            current = current.next
            current_index += 1

        # Out of bounds check
        if current is None:
            print(f"Index {index} is out of bounds!")
            return False

        new_node = Node(data)

        # Case 1: Inserting after the last node (tail)
        if current == self.tail:
            new_node.prev = current
            current.next = new_node
            self.tail = new_node
            return True

        # Case 2: Inserting in between two existing nodes
        new_node.next = current.next
        new_node.prev = current
        current.next.prev = new_node
        current.next = new_node
        return True

    def display_forward(self):
        """Prints the doubly linked list from head to tail."""
        current = self.head

        if not current:
            print("List is empty!")
            return

        while current:
            print(f"{current.data} <-> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
dll = DoublyLinkedList()

# Case 1: Attempting to insert into an empty list
dll.insert_after_index(0, 10)  # Output: List is empty! Cannot insert.

# Populating initial node for testing
dll.head = Node(10)
dll.tail = dll.head

# Case 2: Inserting after index 0 (Tail Insertion)
dll.insert_after_index(0, 30)
dll.display_forward()  # Output: 10 <-> 30 <-> None

# Case 3: Inserting after index 0 (Middle Insertion between 10 and 30)
dll.insert_after_index(0, 20)
dll.display_forward()  # Output: 10 <-> 20 <-> 30 <-> None

# Case 4: Invalid index out of bounds
dll.insert_after_index(10, 99)  # Output: Index 10 is out of bounds!
````

### Conceptual Bridge

* **Tail Insertion (Case 1):** If the target car is the current caboose (`tail`), the new car couples behind it, and the station updates its official caboose tracking marker (`self.tail = new_node`).
* **Middle Insertion (Case 2):** If inserting between two existing cars, 4 coupling adjustments are made in sequence so neither half of the train disconnects.

### Detailed Technical Breakdown

* **`if current == self.tail:`**
Evaluates whether the target address equals the `tail` pointer. If so, updates `self.tail` to maintain constant-time $O(1)$ tail references.
* **Middle Pointer Assignment Order:**
1. `new_node.next = current.next` (Points new node forward to current's successor)
2. `new_node.prev = current` (Points new node backward to current)
3. `current.next.prev = new_node` (Updates successor's backward link to new node)
4. `current.next = new_node` (Updates current's forward link to new node)



> **Execution Critical Rule:** Re-linking `current.next.prev` **must** happen before modifying `current.next`. Overwriting `current.next` first breaks the memory address chain to the right half of the list.

* **Time Complexity:**
* Empty List Check / Validation: $O(1)$ constant time.
* Index Search Traversal: $O(k)$ linear time, where $k$ is the target index (worst-case $O(n)$).
* Pointer Re-linking: $O(1)$ constant time.


* **Space Complexity:** $O(1)$ auxiliary space complexity (allocates a single `Node` instance).

## Accessing node data

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, data):
        """Appends a new node to the end of the doubly linked list."""
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
            return
        self.tail.next = new_node
        new_node.prev = self.tail
        self.tail = new_node

    def get_by_value(self, target_value):
        """Searches for a target value and returns its 0-based index and node data."""
        current = self.head
        index = 0

        # Linear search through memory references
        while current:
            if current.data == target_value:
                return index, current.data  # Returns (index, data) on match
            current = current.next
            index += 1

        return -1, None  # Value not found

    def get_by_index(self, index):
        """Retrieves node data at a specified 0-based index."""
        if index < 0:
            print("Index cannot be negative!")
            return None

        current = self.head
        current_index = 0

        # Traversal until reaching desired index position
        while current:
            if current_index == index:
                return current.data  # Returns data payload on index match
            current = current.next
            current_index += 1

        print(f"Index {index} is out of bounds!")
        return None


# --- Testing the implementation ---
dll = DoublyLinkedList()
dll.append(10)
dll.append(20)
dll.append(30)
dll.append(40)

# Case 1: Accessing Node Data by Value (Found)
idx, val = dll.get_by_value(30)
print(f"Value Search (30): Found at index {idx} with data {val}")
# Output: Value Search (30): Found at index 2 with data 30

# Case 2: Accessing Node Data by Value (Not Found)
idx_not_found, val_not_found = dll.get_by_value(99)
print(f"Value Search (99): Result index {idx_not_found}")
# Output: Value Search (99): Result index -1

# Case 3: Accessing Node Data by Index (Valid Index)
val_at_idx = dll.get_by_index(2)
print(f"Index Access (Index 2): Data is {val_at_idx}")
# Output: Index Access (Index 2): Data is 30

# Case 4: Accessing Node Data by Index (Out of Bounds)
val_out_of_bounds = dll.get_by_index(10)
# Output: Index 10 is out of bounds!
````

## DELETE USER NODE

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, data):
        """Helper method to append nodes for testing."""
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
            return
        self.tail.next = new_node
        new_node.prev = self.tail
        self.tail = new_node

    def delete_by_index(self, index):
        """Deletes a node at the specified 0-based index."""
        if self.head is None:
            print("List is empty! Cannot delete.")
            return None

        if index < 0:
            print("Index cannot be negative!")
            return None

        current = self.head
        current_index = 0

        # Traverse to target index
        while current and current_index < index:
            current = current.next
            current_index += 1

        if current is None:
            print(f"Index {index} is out of bounds!")
            return None

        deleted_data = current.data

        # Case 1: Single-node list
        if self.head == self.tail:
            self.head = None
            self.tail = None
        # Case 2: Deleting head node (index 0)
        elif current == self.head:
            self.head = self.head.next
            self.head.prev = None
        # Case 3: Deleting tail node
        elif current == self.tail:
            self.tail = self.tail.prev
            self.tail.next = None
        # Case 4: Deleting middle node
        else:
            current.prev.next = current.next
            current.next.prev = current.prev

        return deleted_data

    def delete_by_value(self, target_value):
        """Deletes the first node that contains the specified target_value."""
        if self.head is None:
            print("List is empty! Cannot delete.")
            return False

        current = self.head

        # Search for node containing target_value
        while current and current.data != target_value:
            current = current.next

        if current is None:
            print(f"Value '{target_value}' not found in the list!")
            return False

        # Case 1: Single-node list
        if self.head == self.tail:
            self.head = None
            self.tail = None
        # Case 2: Deleting head node
        elif current == self.head:
            self.head = self.head.next
            self.head.prev = None
        # Case 3: Deleting tail node
        elif current == self.tail:
            self.tail = self.tail.prev
            self.tail.next = None
        # Case 4: Deleting middle node
        else:
            current.prev.next = current.next
            current.next.prev = current.prev

        return True

    def display_forward(self):
        """Prints the doubly linked list from head to tail."""
        current = self.head

        if not current:
            print("List is empty!")
            return

        while current:
            print(f"{current.data} <-> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
dll = DoublyLinkedList()

# Pre-populating: 10 <-> 20 <-> 30 <-> 40 <-> 50 <-> None
for val in [10, 20, 30, 40, 50]:
    dll.append(val)

print("Initial List:")
dll.display_forward()

# Case 1: Delete by Index (Middle Node - Index 2, Value 30)
val_deleted = dll.delete_by_index(2)
print(f"\nDeleted index 2 (value: {val_deleted}):")
dll.display_forward()  # Output: 10 <-> 20 <-> 40 <-> 50 <-> None

# Case 2: Delete by Value (Head Node - Value 10)
dll.delete_by_value(10)
print("\nDeleted value 10 (Head):")
dll.display_forward()  # Output: 20 <-> 40 <-> 50 <-> None

# Case 3: Delete by Value (Tail Node - Value 50)
dll.delete_by_value(50)
print("\nDeleted value 50 (Tail):")
dll.display_forward()  # Output: 20 <-> 40 <-> None

# Case 4: Delete by Index Out of Bounds
dll.delete_by_index(10)  # Output: Index 10 is out of bounds!

# Case 5: Delete by Value Not Found
dll.delete_by_value(99)  # Output: Value '99' not found in the list!
````

## DELETE LINKEDLIST

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, data):
        """Helper method to append nodes for testing."""
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
            return
        self.tail.next = new_node
        new_node.prev = self.tail
        self.tail = new_node

    def delete_list(self):
        """Deletes all nodes in the doubly linked list and frees memory references."""
        if self.head is None:
            print("List is already empty!")
            return

        # Iteratively sever forward and backward links to break cyclic references
        current = self.head
        while current:
            next_node = current.next  # Preserve reference to next node
            current.next = None       # Break forward link
            current.prev = None       # Break backward link
            current = next_node       # Advance cursor

        # Reset list boundary entry points
        self.head = None
        self.tail = None
        print("Doubly linked list deleted successfully!")

    def display_forward(self):
        """Prints the doubly linked list from head to tail."""
        current = self.head

        if not current:
            print("List is empty!")
            return

        while current:
            print(f"{current.data} <-> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
dll = DoublyLinkedList()

# Pre-populating list: 10 <-> 20 <-> 30 <-> 40 <-> None
for val in [10, 20, 30, 40]:
    dll.append(val)

print("Initial List:")
dll.display_forward()  # Output: 10 <-> 20 <-> 30 <-> 40 <-> None

# Case 1: Deleting the entire linked list
print("\nDeleting entire linked list...")
dll.delete_list()      # Output: Doubly linked list deleted successfully!

# Case 2: Displaying after deletion
dll.display_forward()  # Output: List is empty!

# Case 3: Attempting to delete an already empty list
dll.delete_list()      # Output: List is already empty!
````
