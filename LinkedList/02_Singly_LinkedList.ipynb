# SINGLY LINKED LIST


## BASIC STRUCTURE FOR CREATING A SINGLY LINKED LIST

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the end of the linked list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = new_node

    def display_linkedlist(self):
        """Prints the linked list contents."""
        current = self.head

        if not current:
            print("List is empty!")
            return  

        while current:
            print(f"{current.data} -> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
ll = LinkedList()

# Case 1: Empty list display
ll.display_linkedlist()  # Output: List is empty!

# Case 2: Populated list display
ll.append(10)
ll.append(20)
ll.append(30)
ll.display_linkedlist()  # Output: 10 -> 20 -> 30 -> None
````

### Detailed Technical Breakdown

* **`new_node = Node(data)`**
**Dynamic Memory Allocation:** Instantiates a new `Node` object in the heap. Memory is allocated dynamically at runtime, setting `new_node.data = data` and `new_node.next = None`.
* **`if self.head is None:`**
**Base Case / Empty List Guard:** Evaluates whether the list contains zero nodes by checking if the head pointer is set to `None`.
* **`self.head = new_node` & `return**`
**$O(1)$ Insertion:** If the list is empty, updates `self.head` to store the address reference of `new_node`. The method exits early via `return`, taking constant time $O(1)$.
* **`current = self.head`**
**Traversal Cursor Initialization:** Creates a temporary pointer variable `current` and initializes it with the memory address held in `self.head`. This allows us to traverse the list without modifying or losing the original `head` reference.
* **`while current.next:`**
**Linear Traversal Condition:** Evaluates the `next` reference field of the node currently pointed to by `current`. The loop continues as long as `current.next` does *not* equal `None` (i.e., another node exists down the chain).
* **`current = current.next`**
**Pointer Dereferencing / Step:** Reassigns the `current` variable to store the memory address of the next node (`current.next`). This advances the traversal cursor forward by one node.
* **`current.next = new_node`**
**Pointer Reassignment / Linking:** Once the loop completes, `current` is pointing to the final node (where `current.next` was `None`). This line updates the final node's `next` field from `None` to hold the memory address of `new_node`, appending it to the structure.

> **Time Complexity:** $O(n)$ — Linear time complexity because the system must traverse all $n$ nodes from the head to locate the tail pointer before inserting.

Advances the pointer to the next memory address. On the final iteration at the tail node, `current.next` is `None`, so `current` becomes `None`, breaking the `while` loop on the next condition check.
* **`print("None")`**
Prints `"None"` at the end of the terminal output stream to explicitly visualize the null pointer termination of the linked list.

> **Time Complexity:** $O(n)$ — Reads each node's memory location exactly once.

## INSERTING A NODE AT THE START

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def prepend(self, data):
        """Inserts a new node at the beginning of the linked list."""
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def display_linkedlist(self):
        """Prints the linked list contents."""
        current = self.head

        if not current:
            print("List is empty!")
            return 

        while current:
            print(f"{current.data} -> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
ll = LinkedList()

# Case 2: Inserting nodes at the start (prepend)
ll.prepend(30)  # List: 30 -> None
ll.prepend(20)  # List: 20 -> 30 -> None
ll.prepend(10)  # List: 10 -> 20 -> 30 -> None

ll.display_linkedlist()  # Output: 10 -> 20 -> 30 -> None
````

* **`new_node = Node(data)`**

* **`new_node.next = self.head`**

## INSERTING A NODE AFTER AN INDEX

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the end of the linked list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = new_node

    def insert_after_index(self, target_index, new_data):
        """Inserts a new node immediately after the specified index."""
        if target_index < 0:
            print(f"Index {target_index} is invalid.")
            return

        if self.head is None:
            print("List is empty! Cannot insert.")
            return

        current = self.head
        current_index = 0

        # Traverse until reaching target_index or the end of the list
        while current and current_index < target_index:
            current = current.next
            current_index += 1

        # If target_index is out of bounds
        if current is None:
            print(f"Index {target_index} is out of bounds.")
            return

        # Insert new node after target node
        new_node = Node(new_data)
        new_node.next = current.next
        current.next = new_node

    def display_linkedlist(self):
        """Prints the linked list contents."""
        current = self.head

        if not current:
            print("List is empty!")
            return  

        while current:
            print(f"{current.data} -> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
ll = LinkedList()

# Populating initial list
ll.append(10)
ll.append(20)
ll.append(30)

# Display initial list
ll.display_linkedlist()
# Output: 10 -> 20 -> 30  -> None

# Inserting 100 after index 1 (after value 20)
ll.insert_after_index(1, 100)

# Display updated list
ll.display_linkedlist()
# Output: 10 -> 20 -> 100 -> 30 -> None
````

## Accessing node data

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the end of the linked list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = new_node

    def access_node_by_value(self, target_data):
        """Searches for a node by value and returns the node object."""
        current = self.head
        position = 0

        while current:
            if current.data == target_data:
                print(f"Value '{target_data}' found at position {position}.")
                return current
            current = current.next
            position += 1

        print(f"Value '{target_data}' not found in the list.")
        return None

    def access_node_by_index(self, target_index):
        """Accesses and returns the data stored at a given index."""
        if target_index < 0:
            print(f"Index {target_index} is invalid.")
            return None

        if self.head is None:
            print("List is empty!")
            return None

        current = self.head
        current_index = 0

        while current:
            if current_index == target_index:
                print(f"Data at index {target_index} is: {current.data}")
                return current.data
            current = current.next
            current_index += 1

        print(
            f"Index {target_index} is out of bounds (List length: {current_index})."
        )
        return None

    def display_linkedlist(self):
        """Prints the linked list contents."""
        current = self.head

        if not current:
            print("List is empty!")
            return 

        while current:
            print(f"{current.data} -> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
ll = LinkedList()

# Populating initial list
ll.append(10)
ll.append(20)
ll.append(30)
ll.append(90)
ll.append(60)
ll.append(50)

# Case 1: Access node by value
ll.access_node_by_value(90)
# Output: Value '90' found at position 3.

# Case 2: Access node by index
ll.access_node_by_index(5)
# Output: Data at index 5 is: 50

# Display full list
ll.display_linkedlist()
# Output: 10 -> 20 -> 30 -> 90 -> 60 -> 50 -> None
````

## DELETE FIRST NODE

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the end of the linked list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = new_node

    def delete_first(self):
        """Removes the first node (head) from the linked list."""
        if self.head is None:
            print("List is empty! Nothing to delete.")
            return

        # Shift head pointer to the second node
        self.head = self.head.next

    def display_linkedlist(self):
        """Prints the linked list contents."""
        current = self.head

        if not current:
            print("List is empty!")
            return

        while current:
            print(f"{current.data} -> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
ll = LinkedList()

# Populating initial list
ll.append(10)
ll.append(20)
ll.append(30)
ll.append(90)
ll.append(60)
ll.append(50)

# Display initial list
ll.display_linkedlist()
# Output: 10 -> 20 -> 30 -> 90 -> 60 -> 50 -> None

# Deleting the first node (10)
ll.delete_first()

# Display updated list
ll.display_linkedlist()
# Output: 20 -> 30 -> 90 -> 60 -> 50 -> None
````


### Detailed Breakdown of `delete_first`

```python
    def delete_first(self):
        if self.head is None:
            print("List is empty! Nothing to delete.")
            return

```

* **`if self.head is None:` & `return**`
**Null Pointer Guard:** Verifies if the list is already empty before attempting deletion. If `self.head` is `None`, attempting to access `self.head.next` would trigger an `AttributeError`.

```python
        self.head = self.head.next

```

* **`self.head = self.head.next`**
**Pointer Shift:** Dereferences the current head node (`self.head`) to retrieve its `next` field (the address of the second node), then updates `self.head` to point directly to that second node.
* **Automatic Memory Deallocation (Garbage Collection):**
Once `self.head` stops referencing the original first node (`Node(10)`), its reference count drops to zero. Python’s internal **Garbage Collector** automatically frees that memory location.

> **Time Complexity:** **$O(1)$ Constant Time** — No list traversal is required. Updating the entry pointer executes in a single memory operation regardless of list size.

## DELETE LAST NODE

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the end of the linked list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = new_node

    def delete_last(self):
        """Removes the last node (tail) from the linked list."""
        if self.head is None:
            print("List is empty! Nothing to delete.")
            return

        # Case 1: Single node in the list
        if self.head.next is None:
            self.head = None
            return

        # Case 2: Traverse to find the second-to-last (penultimate) node
        current = self.head
        while current.next.next:
            current = current.next

        # Disconnect the link to the last node
        current.next = None

    def display_linkedlist(self):
        """Prints the linked list contents."""
        current = self.head

        if not current:
            print("List is empty!")
            return  

        while current:
            print(f"{current.data} -> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
ll = LinkedList()

# Populating initial list
ll.append(10)
ll.append(20)
ll.append(30)
ll.append(90)
ll.append(60)
ll.append(50)

# Display initial list
ll.display_linkedlist()
# Output: 10 -> 20 -> 30 -> 90 -> 60 -> 50 -> None

# Deleting the last node (50)
ll.delete_last()

# Display updated list
ll.display_linkedlist()
# Output: 10 -> 20 -> 30 -> 90 -> 60 -> None
````


### Detailed Breakdown of `delete_last`

```python
    def delete_last(self):
        if self.head is None:
            print("List is empty! Nothing to delete.")
            return

```

* **`if self.head is None:` & `return**`
**Empty List Guard:** Verifies if the list contains nodes. If empty, terminates immediately to avoid null pointer dereferencing.

```python
        if self.head.next is None:
            self.head = None
            return

```

* **`if self.head.next is None:`**
**Single-Node Guard:** Evaluates if `head` is the only node present. If `self.head.next` is `None`, setting `self.head = None` empties the list in $O(1)$ time.

```python
        current = self.head
        while current.next.next:
            current = current.next

```

* **`while current.next.next:`**
**Look-Ahead Traversal:** Instead of stopping at the last node (`current.next`), the loop inspects **two steps ahead** (`current.next.next`).
* When `current.next.next` evaluates to `None`, `current` is standing at the **second-to-last node**.



```python
        current.next = None

```

* **`current.next = None`**
**Disconnection & Garbage Collection:** Reassigns the second-to-last node's `next` pointer from the tail node's address to `None`.
* The former tail node loses all pointer references, dropping its reference count to zero so Python's **Garbage Collector** safely deallocates its memory.



> **Time Complexity:** **$O(n)$ Linear Time** — Requires traversing through $n - 1$ nodes from `head` to locate the penultimate node before unlinking.

## DELETE USER NODE

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the end of the linked list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = new_node

    def delete_value(self, target_data):
        """Deletes the first node matching target_data."""
        if self.head is None:
            print("List is empty! Nothing to delete.")
            return

        # Case 1: Target node is the head
        if self.head.data == target_data:
            self.head = self.head.next
            return

        # Case 2: Traverse to find the node BEFORE the target node
        current = self.head
        while current.next and current.next.data != target_data:
            current = current.next

        if current.next is None:
            print(f"Value '{target_data}' not found in list.")
            return

        # Bypass the target node to un-link it
        current.next = current.next.next

    def delete_at_index(self, target_index):
        """Deletes the node at the specified 0-based index offset."""
        if target_index < 0:
            print(f"Index {target_index} is invalid.")
            return

        if self.head is None:
            print("List is empty! Nothing to delete.")
            return

        # Case 1: Target node is at index 0 (head)
        if target_index == 0:
            self.head = self.head.next
            return

        # Case 2: Traverse to the node BEFORE target_index (target_index - 1)
        current = self.head
        current_index = 0

        while current.next and current_index < target_index - 1:
            current = current.next
            current_index += 1

        # Out of bounds check
        if current.next is None:
            print(f"Index {target_index} is out of bounds.")
            return

        # Bypass the target node to un-link it
        current.next = current.next.next

    def display_linkedlist(self):
        """Prints the linked list contents."""
        current = self.head

        if not current:
            print("List is empty!")
            return

        while current:
            print(f"{current.data} -> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
ll = LinkedList()

# Populating initial list
ll.append(10)
ll.append(20)
ll.append(30)
ll.append(90)
ll.append(60)
ll.append(50)

print("Original List:")
ll.display_linkedlist()
# Output: 10 -> 20 -> 30 -> 90 -> 60 -> 50 -> None

print("\nDeleting Value 20:")
ll.delete_value(20)
ll.display_linkedlist()
# Output: 10 -> 30 -> 90 -> 60 -> 50 -> None

print("\nDeleting Node at Index 3:")
ll.delete_at_index(3)
ll.display_linkedlist()
# Output: 10 -> 30 -> 90 -> 50 -> None
````


### Detailed Breakdown of Deletion Methods

#### 1. Delete by Value (`delete_value`)


* **`if self.head.data == target_data:`**
**Head Special Case:** If the first node holds `target_data`, reassigns `self.head` to `self.head.next` in $O(1)$ time.
* **`while current.next and current.next.data != target_data:`**
**Predecessor Look-Ahead:** Advances `current` as long as the *next* node's data is not the target. This ensures traversal halts directly at the node **before** the target.
* **`current.next = current.next.next`**
**Bypass:** Overwrites `current.next` to hold the address of the node *after* the target, dropping all references to the target node for Python garbage collection.

> **Time Complexity:** **$O(n)$ Linear Time** — Requires searching through up to $n$ nodes to find the matching value.

* **`if target_index == 0:`**
**Head Index Case:** Directly shifts `self.head` pointer to the second node.
* **`while current.next and current_index < target_index - 1:`**
**Offset Look-Ahead:** Traverses until `current_index` reaches `target_index - 1`. Halting at index $k-1$ is mandatory because single-linked list nodes cannot look backward.
* **`if current.next is None:`**
**Out-of-Bounds Check:** Triggers if `target_index` exceeds the valid node range.
* **`current.next = current.next.next`**
**Bypass:** Bridges the link over the node at `target_index`, unlinking it from the chain.

> **Time Complexity:** **$O(k)$ Linear Time** (where $k = \text{target\_index}$) — Traversing to index $k-1$ requires $k$ pointer evaluations.

## DELETE LINKEDLIST

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the end of the linked list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = new_node

    def delete_entire_list(self):
        """Deletes all nodes in the linked list by clearing the head reference."""
        self.head = None
        print("Entire linked list has been deleted.")

    def display_linkedlist(self):
        """Prints the linked list contents."""
        current = self.head

        if not current:
            print("List is empty!")
            return

        while current:
            print(f"{current.data} -> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
ll = LinkedList()

# Populating initial list
ll.append(10)
ll.append(20)
ll.append(30)
ll.append(90)
ll.append(60)
ll.append(50)

print("Original List:")
ll.display_linkedlist()
# Output: 10 -> 20 -> 30 -> 90 -> 60 -> 50 -> None

print("\nDeleting List:")
ll.delete_entire_list()

print("\nDisplaying List Post-Deletion:")
ll.display_linkedlist()
# Output: List is empty!
````


### Detailed Breakdown of `delete_entire_list`

```python
    def delete_entire_list(self):
        self.head = None
        print("Entire linked list has been deleted.")

```

* **`self.head = None`**
**Root Pointer Reset:** Reassigns the `head` pointer to `None`. Because a singly linked list is navigated strictly top-down from `head`, clearing this reference severs all access to the nodes in the list.
* **Cascading Garbage Collection in Python:**
1. Once `self.head` is set to `None`, `Node(10)` has zero active references pointing to it.
2. Python's Garbage Collector deallocates `Node(10)`.
3. Deallocating `Node(10)` removes the reference to `Node(20)`, causing its reference count to fall to zero as well.
4. This creates a **cascading reference collapse**, freeing all nodes in the chain automatically without explicit traversal loop requirements.



> **Time Complexity:** **$O(1)$ Constant Time** (in Python) — Clearing `self.head` is a single pointer assignment operation. (Language-level deallocation of the $n$ orphan nodes is handled asynchronously by the Python interpreter).
