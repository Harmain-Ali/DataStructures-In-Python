# LINKED LIST IN PYTHON



## 1. What is a Linked List Data Structure?

A **Linked List** is a linear data structure where elements are not stored in contiguous (adjacent) memory locations like arrays. Instead, each element—called a **node**—contains two primary parts:

* **Data:** The actual value or payload stored in the element.
* **Pointer/Reference:** A link pointing to the memory address of the next node in the sequence.

The list starts at a reference point called the **Head** and terminates at a node pointing to `NULL` (or `None`), indicating the end of the list.

```
[Head] -> [ Data | Next ] -> [ Data | Next ] -> [ Data | NULL ]

```

---

## 2. Types of Linked Lists

### Singly Linked List

A basic linked list where navigation is strictly one-way (forward). Each node stores its data and a single pointer pointing to the next node in the sequence. The last node points to `NULL`.

### Doubly Linked List

A linked list where nodes can be traversed in both forward and backward directions. Each node contains three components: data, a pointer to the **next** node, and a pointer to the **previous** node.

### Circular Linked List

A variation where the last node does not point to `NULL`. Instead, its `next` pointer points back to the **Head** (first node), forming a continuous loop. This can be built using either singly or doubly linked list nodes.

### Circular Doubly Linked List

A combination of circular and doubly linked lists. The last node's `next` pointer points to the head, and the head node's `previous` pointer points to the last node, enabling circular traversal in both directions.

---

## 3. Why Linked List? (Advantages over Arrays)

| Feature | Linked List | Array |
| --- | --- | --- |
| **Dynamic Memory Allocation** | Memory is allocated at runtime as elements are added. No fixed size required. | Size must be defined ahead of time in contiguous memory. |
| **Efficient Insertion/Deletion** | $O(1)$ constant-time insertion and deletion at a known pointer position without shifting elements. | $O(n)$ linear time due to the need to shift remaining elements in memory. |
| **Memory Efficiency** | Uses exact memory needed for current elements (plus pointer overhead). | Can lead to wasted memory if allocated capacity is underutilized. |

---

## 4. Real-World Use Cases for Every Linked List Type

### 1. Singly Linked List

* **Undo Functionality / Call Stacks:** Implementing dynamic stacks where function execution context or state steps are continuously pushed and popped.
* **Symbol Tables in Compilers:** Compilers frequently use singly linked lists to build dynamic lookup tables for variables and functions defined within execution scopes.
* **Image Viewer Next Button:** Simple sequential media workflows that only require progressing forward.

### 2. Doubly Linked List

* **Browser Navigation:** The "Back" and "Forward" buttons in web browsers use a doubly linked list to navigate through your browsing history seamlessly in both directions.
* **LRU Cache (Least Recently Used):** High-performance caching systems combine a hash map with a doubly linked list to achieve $O(1)$ node removal and insertion when evicting stale data.
* **Undo/Redo Actions in Software:** Applications like Photoshop or Word processors use bidirectional references to move backward (Undo) and forward (Redo) through edit histories.

### 3. Circular Linked List

* **Round-Robin CPU Scheduling:** Operating systems cycle through active processes by giving each a time slice before looping back to the first process in the queue.
* **Multiplayer Turn-Based Games:** Games like Monopoly or UNO loop continuously through players in fixed order; once the last player finishes, the turn returns to the first player.
* **Media Player Playlists (Repeat Mode):** Playlists configured to continuously play audio tracks in an infinite loop without stopping at the end of the list.

### 4. Circular Doubly Linked List

* **Carousel UI Components:** Frontend image or card carousels that allow infinite continuous scrolling both left and right without reaching a hard border.
* **Fibonacci Heaps & Advanced Graph Algorithms:** Used internally in complex algorithmic structures where nodes need rapid insertion, deletion, and continuous multi-directional merging.
