# LINKED LIST IN PYTHON



## 1. What is a Linked List Data Structure?

A **Linked List** is a linear data structure where elements are not stored in contiguous (adjacent) memory locations like arrays. Instead, each element—called a **node**—contains two primary parts:

* **Data:** The actual value or payload stored in the element.
* **Pointer/Reference:** A link pointing to the memory address of the next node in the sequence.

The list starts at a reference point called the **Head** and terminates at a node pointing to `NULL` (or `None`), indicating the end of the list.

```
[Head] -> [ Data | Next ] -> [ Data | Next ] -> [ Data | NULL ]

```

---

## 2. Types of Linked Lists

### Singly Linked List

A basic linked list where navigation is strictly one-way (forward). Each node stores its data and a single pointer pointing to the next node in the sequence. The last node points to `NULL`.

### Doubly Linked List

A linked list where nodes can be traversed in both forward and backward directions. Each node contains three components: data, a pointer to the **next** node, and a pointer to the **previous** node.

### Circular Linked List

A variation where the last node does not point to `NULL`. Instead, its `next` pointer points back to the **Head** (first node), forming a continuous loop. This can be built using either singly or doubly linked list nodes.

### Circular Doubly Linked List

A combination of circular and doubly linked lists. The last node's `next` pointer points to the head, and the head node's `previous` pointer points to the last node, enabling circular traversal in both directions.

---

## 3. Why Linked List? (Advantages over Arrays)

| Feature | Linked List | Array |
| --- | --- | --- |
| **Dynamic Memory Allocation** | Memory is allocated at runtime as elements are added. No fixed size required. | Size must be defined ahead of time in contiguous memory. |
| **Efficient Insertion/Deletion** | $O(1)$ constant-time insertion and deletion at a known pointer position without shifting elements. | $O(n)$ linear time due to the need to shift remaining elements in memory. |
| **Memory Efficiency** | Uses exact memory needed for current elements (plus pointer overhead). | Can lead to wasted memory if allocated capacity is underutilized. |

---

## 4. Real-World Use Cases for Every Linked List Type

### 1. Singly Linked List

* **Undo Functionality / Call Stacks:** Implementing dynamic stacks where function execution context or state steps are continuously pushed and popped.
* **Symbol Tables in Compilers:** Compilers frequently use singly linked lists to build dynamic lookup tables for variables and functions defined within execution scopes.
* **Image Viewer Next Button:** Simple sequential media workflows that only require progressing forward.

### 2. Doubly Linked List

* **Browser Navigation:** The "Back" and "Forward" buttons in web browsers use a doubly linked list to navigate through your browsing history seamlessly in both directions.
* **LRU Cache (Least Recently Used):** High-performance caching systems combine a hash map with a doubly linked list to achieve $O(1)$ node removal and insertion when evicting stale data.
* **Undo/Redo Actions in Software:** Applications like Photoshop or Word processors use bidirectional references to move backward (Undo) and forward (Redo) through edit histories.

### 3. Circular Linked List

* **Round-Robin CPU Scheduling:** Operating systems cycle through active processes by giving each a time slice before looping back to the first process in the queue.
* **Multiplayer Turn-Based Games:** Games like Monopoly or UNO loop continuously through players in fixed order; once the last player finishes, the turn returns to the first player.
* **Media Player Playlists (Repeat Mode):** Playlists configured to continuously play audio tracks in an infinite loop without stopping at the end of the list.

### 4. Circular Doubly Linked List

* **Carousel UI Components:** Frontend image or card carousels that allow infinite continuous scrolling both left and right without reaching a hard border.
* **Fibonacci Heaps & Advanced Graph Algorithms:** Used internally in complex algorithmic structures where nodes need rapid insertion, deletion, and continuous multi-directional merging.

# SINGLY LINKED LIST


## BASIC STRUCTURE FOR CREATING A SINGLY LINKED LIST

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the end of the linked list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = new_node

    def display_linkedlist(self):
        """Prints the linked list contents."""
        current = self.head

        if not current:
            print("List is empty!")
            return  

        while current:
            print(f"{current.data} -> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
ll = LinkedList()

# Case 1: Empty list display
ll.display_linkedlist()  # Output: List is empty!

# Case 2: Populated list display
ll.append(10)
ll.append(20)
ll.append(30)
ll.display_linkedlist()  # Output: 10 -> 20 -> 30 -> None
````

### Detailed Technical Breakdown

* **`new_node = Node(data)`**
**Dynamic Memory Allocation:** Instantiates a new `Node` object in the heap. Memory is allocated dynamically at runtime, setting `new_node.data = data` and `new_node.next = None`.
* **`if self.head is None:`**
**Base Case / Empty List Guard:** Evaluates whether the list contains zero nodes by checking if the head pointer is set to `None`.
* **`self.head = new_node` & `return**`
**$O(1)$ Insertion:** If the list is empty, updates `self.head` to store the address reference of `new_node`. The method exits early via `return`, taking constant time $O(1)$.
* **`current = self.head`**
**Traversal Cursor Initialization:** Creates a temporary pointer variable `current` and initializes it with the memory address held in `self.head`. This allows us to traverse the list without modifying or losing the original `head` reference.
* **`while current.next:`**
**Linear Traversal Condition:** Evaluates the `next` reference field of the node currently pointed to by `current`. The loop continues as long as `current.next` does *not* equal `None` (i.e., another node exists down the chain).
* **`current = current.next`**
**Pointer Dereferencing / Step:** Reassigns the `current` variable to store the memory address of the next node (`current.next`). This advances the traversal cursor forward by one node.
* **`current.next = new_node`**
**Pointer Reassignment / Linking:** Once the loop completes, `current` is pointing to the final node (where `current.next` was `None`). This line updates the final node's `next` field from `None` to hold the memory address of `new_node`, appending it to the structure.

> **Time Complexity:** $O(n)$ — Linear time complexity because the system must traverse all $n$ nodes from the head to locate the tail pointer before inserting.

### Detailed Technical Breakdown

* **`current = self.head`**
Initializes a temporary traversal pointer `current` at the head reference of the list.
* **`if not current:` & `return**`
**Null Check Guard:** Evaluates if `current` is `None` (empty list). If true, it prints a message and terminates execution via `return` to avoid unnecessary processing or potential runtime errors.
* **`while current:`**
**Iterative Traversal Loop:** Unlike `while current.next`, this condition checks if `current` itself holds a valid node address. The loop executes for *every* node, including the tail node.
* **`print(f"{current.data} -> ", end="")`**
**Data Extraction:** Dereferences the `current` pointer to read `current.data` and prints it to stdout. `end=""` suppresses the default newline character to keep the formatted chain output on a single line.
* **`current = current.next`**
Advances the pointer to the next memory address. On the final iteration at the tail node, `current.next` is `None`, so `current` becomes `None`, breaking the `while` loop on the next condition check.
* **`print("None")`**
Prints `"None"` at the end of the terminal output stream to explicitly visualize the null pointer termination of the linked list.

> **Time Complexity:** $O(n)$ — Reads each node's memory location exactly once.

## INSERTING A NODE AT THE START

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def prepend(self, data):
        """Inserts a new node at the beginning of the linked list."""
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def display_linkedlist(self):
        """Prints the linked list contents."""
        current = self.head

        if not current:
            print("List is empty!")
            return 

        while current:
            print(f"{current.data} -> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
ll = LinkedList()

# Case 2: Inserting nodes at the start (prepend)
ll.prepend(30)  # List: 30 -> None
ll.prepend(20)  # List: 20 -> 30 -> None
ll.prepend(10)  # List: 10 -> 20 -> 30 -> None

ll.display_linkedlist()  # Output: 10 -> 20 -> 30 -> None
````

## INSERTING A NODE AFTER AN INDEX

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the end of the linked list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = new_node

    def insert_after_index(self, target_index, new_data):
        """Inserts a new node immediately after the specified index."""
        if target_index < 0:
            print(f"Index {target_index} is invalid.")
            return

        if self.head is None:
            print("List is empty! Cannot insert.")
            return

        current = self.head
        current_index = 0

        # Traverse until reaching target_index or the end of the list
        while current and current_index < target_index:
            current = current.next
            current_index += 1

        # If target_index is out of bounds
        if current is None:
            print(f"Index {target_index} is out of bounds.")
            return

        # Insert new node after target node
        new_node = Node(new_data)
        new_node.next = current.next
        current.next = new_node

    def display_linkedlist(self):
        """Prints the linked list contents."""
        current = self.head

        if not current:
            print("List is empty!")
            return  

        while current:
            print(f"{current.data} -> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
ll = LinkedList()

# Populating initial list
ll.append(10)
ll.append(20)
ll.append(30)

# Display initial list
ll.display_linkedlist()
# Output: 10 -> 20 -> 30  -> None

# Inserting 100 after index 1 (after value 20)
ll.insert_after_index(1, 100)

# Display updated list
ll.display_linkedlist()
# Output: 10 -> 20 -> 100 -> 30 -> None
````

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the end of the linked list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = new_node

    def access_node_by_value(self, target_data):
        """Searches for a node by value and returns the node object."""
        current = self.head
        position = 0

        while current:
            if current.data == target_data:
                print(f"Value '{target_data}' found at position {position}.")
                return current
            current = current.next
            position += 1

        print(f"Value '{target_data}' not found in the list.")
        return None

    def access_node_by_index(self, target_index):
        """Accesses and returns the data stored at a given index."""
        if target_index < 0:
            print(f"Index {target_index} is invalid.")
            return None

        if self.head is None:
            print("List is empty!")
            return None

        current = self.head
        current_index = 0

        while current:
            if current_index == target_index:
                print(f"Data at index {target_index} is: {current.data}")
                return current.data
            current = current.next
            current_index += 1

        print(
            f"Index {target_index} is out of bounds (List length: {current_index})."
        )
        return None

    def display_linkedlist(self):
        """Prints the linked list contents."""
        current = self.head

        if not current:
            print("List is empty!")
            return 

        while current:
            print(f"{current.data} -> ", end="")
            current = current.next
        print("None")


# --- Testing the implementation ---
ll = LinkedList()

# Populating initial list
ll.append(10)
ll.append(20)
ll.append(30)
ll.append(90)
ll.append(60)
ll.append(50)

# Case 1: Access node by value
ll.access_node_by_value(90)
# Output: Value '90' found at position 3.

# Case 2: Access node by index
ll.access_node_by_index(5)
# Output: Data at index 5 is: 50

# Display full list
ll.display_linkedlist()
# Output: 10 -> 20 -> 30 -> 90 -> 60 -> 50 -> None
````

### Detailed Breakdown of Access Methods

#### 1. Search by Value (`access_node_by_value`)

```python
    def access_node_by_value(self, target_data):

        position = 0

        while current:

        return None

```

* **`current = self.head` & `position = 0**`
Initializes the traversal pointer at the head node and sets the positional tracking counter to zero.
* **`while current:`**
Iterates through memory as long as `current` references a valid node.
* **`if current.data == target_data:`**
Compares the payload of the active node with `target_data`. If matched, prints confirmation and returns the memory reference of the node.
* **`current = current.next` & `position += 1**`
Advances the pointer to the adjacent memory reference and increments the positional counter.
* **`return None`**
Triggers if the entire list is traversed without finding `target_data`.

> **Time Complexity:** **$O(n)$ Linear Time** — In the worst-case scenario (item at the end or not present), every node in the list must be inspected.

#### 2. Search by Index (`access_node_by_index`)


* **`if target_index < 0:` & `if self.head is None:**`
**Guards:** Validates input bound sanity and ensures operations do not evaluate on an empty list reference.
* **`if current_index == target_index:`**
Checks if the current traversal depth matches the requested offset index.
* **`return current.data`**
Retrieves and returns the stored payload directly once the target offset is reached.
* **`print(f"Index {target_index} is out of bounds...")`**
Triggers if `target_index` exceeds the total node count of the list.

> **Time Complexity:** **$O(k)$ Linear Time** (where $k = \text{target\_index}$) — Unlike arrays with $O(1)$ random access, linked lists must sequentially traverse pointers step-by-step to reach an index.
