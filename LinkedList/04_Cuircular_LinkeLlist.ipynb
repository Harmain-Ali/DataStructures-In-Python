# CUIRCULAR SINGLY LINKED LIST

## CREATING A CUIRCULAR SINGLY LINKED LIST

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the circular linked list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            new_node.next = self.head
            return

        current = self.head
        while current.next != self.head:
            current = current.next

        current.next = new_node
        new_node.next = self.head

    def display_linkedlist(self):
        """Prints the circular linked list contents."""
        if self.head is None:
            print("List is empty!")
            return

        current = self.head
        while True:
            print(f"{current.data} -> ", end="")
            current = current.next
            if current == self.head:
                break
        print("(HEAD)")


# --- Testing the implementation ---
cll = CircularLinkedList()

# Case 1: Empty list display
cll.display_linkedlist()  # Output: List is empty!

# Case 2: Populated list display
cll.append(10)
cll.append(20)
cll.append(30)
cll.display_linkedlist()  # Output: 10 -> 20 -> 30 -> (HEAD)
````

### Detailed Technical Breakdown

* **`new_node = Node(data)`**
**Heap Allocation:** Dynamically allocates dynamic memory for a new node with data value `data`.
* **`if self.head is None:`**
**First Node Boundary Condition:** Checks if the circular list contains zero elements.
* **`self.head = new_node` & `new_node.next = self.head**`
**Self-Referential Linking:** When inserting into an empty list, `self.head` points to `new_node`, and `new_node.next` is assigned `self.head` (pointing to itself). This forms the smallest valid circle ($O(1)$ time complexity).
* **`current = self.head`**
**Traversal Initializer:** Sets `current` pointer to `self.head` to scan for the current terminal node.
* **`while current.next != self.head:`**
**Circular Terminal Condition:** Traverses until locating the node whose `next` pointer references `self.head` (i.e., the current tail node).
* **`current = current.next`**
Advances pointer forward across the list.
* **`current.next = new_node` & `new_node.next = self.head**`
**Ring Reconstruction:** Reroutes the former tail's `next` pointer to `new_node`, and links `new_node.next` back to `self.head`. This maintains a uninterrupted ring.

> **Time Complexity:** $O(n)$ — Traverses through all $n$ nodes to locate the tail before appending.

### INSERTING A NODE AT THE START


````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:
    def __init__(self):
        self.head = None

    def prepend(self, data):
        """Inserts a new node at the beginning (head) of the circular linked list."""
        new_node = Node(data)

        # Case 1: Empty list
        if self.head is None:
            self.head = new_node
            new_node.next = self.head
            return

        # Case 2: Populated list - traverse to find the last node (tail)
        current = self.head
        while current.next != self.head:
            current = current.next

        # Link new node to the current head
        new_node.next = self.head
        # Re-route tail node's pointer to the new node
        current.next = new_node
        # Update head reference to the new node
        self.head = new_node

    def display_linkedlist(self):
        """Prints the circular linked list contents."""
        if self.head is None:
            print("List is empty!")
            return

        current = self.head
        while True:
            print(f"{current.data} -> ", end="")
            current = current.next
            if current == self.head:
                break
        print("(HEAD)")


# --- Testing the implementation ---
cll = CircularLinkedList()

# Case 1: Prepending to an empty list
cll.prepend(10)
cll.display_linkedlist()  # Output: 10 -> (HEAD)

# Case 2: Prepending to a populated list
cll.prepend(5)
cll.display_linkedlist()  # Output: 5 -> 10 -> (HEAD)

cll.prepend(1)
cll.display_linkedlist()  # Output: 1 -> 5 -> 10 -> (HEAD)
````

**Dynamic Allocation:** Allocates heap space for a new `Node` instance storing the parameter `data`.
* **`if self.head is None:`**
**Empty Boundary Check:** Checks whether the structure currently contains zero elements.
* **`self.head = new_node` & `new_node.next = self.head**`
**Single Node Circular Reference:** When inserting into an empty list, the new node's `next` pointer loops back onto itself, and `self.head` is set to this single node ($O(1)$ base case).
* **`current = self.head` & `while current.next != self.head:**`
**Tail Search Traversal:** Searches for the current tail node (the node whose `next` pointer holds the address of `self.head`). In a standard singly circular list with only a `head` reference, locating the tail requires traversing the entire structure.
* **`new_node.next = self.head`**
**Forward Linkage:** Connects the new node's forward reference to the *current* head node before `self.head` is overwritten.
* **`current.next = new_node`**
**Tail Linkage Rerouting:** Updates the tail node's `next` pointer from the old head address to the dynamic heap address of `new_node`.
* **`self.head = new_node`**
**Head Update:** Moves the `self.head` pointer to reference `new_node`. `new_node` is now officially the start node of the circular list.

> **Time Complexity:** $O(n)$ — Linear time complexity because locating the tail node to update its pointer requires scanning all $n$ nodes. *(Note: If the class maintained a dedicated `tail` reference pointer, this operation could be optimized to $O(1)$).*

## INSERTING A NODE AFTER AN INDEX

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the end of the circular linked list."""
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            new_node.next = self.head
            return

        current = self.head
        while current.next != self.head:
            current = current.next

        current.next = new_node
        new_node.next = self.head

    def insert_after_index(self, index, data):
        """Inserts a new node immediately after the specified 0-based index."""
        if self.head is None:
            print("List is empty! Cannot insert.")
            return False

        if index < 0:
            print("Index cannot be negative!")
            return False

        current = self.head
        curr_index = 0

        # Traversal loop to locate the node at the specified index
        while True:
            if curr_index == index:
                # Found target node; allocate and re-route references
                new_node = Node(data)
                new_node.next = current.next
                current.next = new_node
                return True

            current = current.next
            curr_index += 1

            # Termination guard: looped full circle without finding the index
            if current == self.head:
                print(f"Index {index} is out of bounds!")
                return False

    def display_linkedlist(self):
        """Prints the circular linked list contents."""
        if self.head is None:
            print("List is empty!")
            return

        current = self.head
        while True:
            print(f"{current.data} -> ", end="")
            current = current.next
            if current == self.head:
                break
        print("(HEAD)")


# --- Testing the implementation ---
cll = CircularLinkedList()

# Populate initial list: 10 -> 20 -> 30 -> (HEAD)
cll.append(10)
cll.append(20)
cll.append(30)
print("Initial List:")
cll.display_linkedlist()  # Output: 10 -> 20 -> 30 -> (HEAD)

# Case 1: Insert after index 0 (between 10 and 20)
cll.insert_after_index(0, 15)
cll.display_linkedlist()  # Output: 10 -> 15 -> 20 -> 30 -> (HEAD)

# Case 2: Insert after last index (index 3, which holds data 30)
cll.insert_after_index(3, 40)
cll.display_linkedlist()  # Output: 10 -> 15 -> 20 -> 30 -> 40 -> (HEAD)

# Case 3: Out of bounds insertion attempt
cll.insert_after_index(10, 99)  # Output: Index 10 is out of bounds!
````

### Detailed Technical Breakdown

* **`if self.head is None:` & `if index < 0:**`
**Initial Boundary Guards:** Validates that the list contains data and that the input index is non-negative before allocating memory or initializing counters.
* **`current = self.head` & `curr_index = 0**`

* **`if curr_index == index:`**

* **`new_node.next = current.next`**
**Step 1 of Reference Re-routing:** Assigns `new_node.next` to point to the node currently following `current`. **Crucial:** This step must happen before changing `current.next` to avoid losing the address reference to the rest of the list.
* **`current.next = new_node`**
**Step 2 of Reference Re-routing:** Reassigns `current.next` to store the address of `new_node`.
* **`current = current.next` & `curr_index += 1**`
**Pointer Dereferencing & Step Increment:** Moves `current` to the next address location and increments the numerical counter.
* **`if current == self.head:`**
**Out of Bounds Cycle Detection:** If `current` wraps around to `self.head` before finding `curr_index == index`, the index exceeds the list length ($index \ge n$). The method alerts the user and returns `False`.

> **Time Complexity:** $O(k)$ where $k$ is the target index (bounded by $O(n)$ worst-case time complexity when inserting after the tail node or checking for out-of-bounds indices).

## Accessing node data

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the end of the circular linked list."""
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            new_node.next = self.head
            return

        current = self.head
        while current.next != self.head:
            current = current.next

        current.next = new_node
        new_node.next = self.head

    def access_by_index(self, index):
        """Retrieves and returns the data stored at a specific 0-based index."""
        if self.head is None:
            print("Error: List is empty!")
            return None

        if index < 0:
            print("Error: Index cannot be negative!")
            return None

        current = self.head
        curr_index = 0

        while True:
            # Check if current node matches the target index
            if curr_index == index:
                return current.data

            current = current.next
            curr_index += 1

            # Full cycle check: returned to head without finding index
            if current == self.head:
                print(f"Error: Index {index} is out of bounds!")
                return None

    def access_by_value(self, target_value):
        """Searches for a node by its stored value and returns its 0-based index position."""
        if self.head is None:
            print("Error: List is empty!")
            return -1

        current = self.head
        curr_index = 0

        while True:
            # Check if current node holds the target value
            if current.data == target_value:
                return curr_index  # Returns index location of the payload

            current = current.next
            curr_index += 1

            # Full cycle check: returned to head without finding value
            if current == self.head:
                print(f"Error: Value '{target_value}' not found in the list!")
                return -1

    def display_linkedlist(self):
        """Prints the circular linked list contents."""
        if self.head is None:
            print("List is empty!")
            return

        current = self.head
        while True:
            print(f"{current.data} -> ", end="")
            current = current.next
            if current == self.head:
                break
        print("(HEAD)")


# --- Testing the implementation ---
cll = CircularLinkedList()

# Populate list: 10 -> 20 -> 30 -> 40 -> (HEAD)
cll.append(10)
cll.append(20)
cll.append(30)
cll.append(40)

print("Current List Structure:")
cll.display_linkedlist()

# --- 1. Accessing Data By Index ---
print("\n--- ACCESS BY INDEX ---")
val_at_idx_2 = cll.access_by_index(2)
print(f"Data at Index 2: {val_at_idx_2}")  # Output: 30

val_at_idx_6 = cll.access_by_index(6)
print(f"Data at Index 6: {val_at_idx_6}")  # Output: Error + None

# --- 2. Accessing Data By Value ---
print("\n--- ACCESS BY VALUE ---")
idx_of_30 = cll.access_by_value(30)
print(f"Value 30 found at Index: {idx_of_30}")  # Output: 2

idx_of_99 = cll.access_by_value(99)
print(f"Value 99 found at Index: {idx_of_99}")  # Output: Error + -1
````

## SEARCHING LINKEDLIST

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the end of the circular linked list."""
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            new_node.next = self.head
            return

        current = self.head
        while current.next != self.head:
            current = current.next

        current.next = new_node
        new_node.next = self.head

    def find_index_of_node(self, target_value):
        """Searches for target_value and returns its 0-based index position.

        Returns -1 if the required element is not found.
        """
        if self.head is None:
            print("Error: List is empty!")
            return -1

        current = self.head
        curr_index = 0

        while True:
            # Check if current node's data payload matches the target value
            if current.data == target_value:
                return curr_index

            current = current.next
            curr_index += 1

            # Termination condition: returned back to head after scanning all nodes
            if current == self.head:
                break

        print(f"Element '{target_value}' not found in the list!")
        return -1

    def display_linkedlist(self):
        """Prints the circular linked list contents."""
        if self.head is None:
            print("List is empty!")
            return

        current = self.head
        while True:
            print(f"{current.data} -> ", end="")
            current = current.next
            if current == self.head:
                break
        print("(HEAD)")


# --- Testing the implementation ---
cll = CircularLinkedList()

# Populate list: 10 -> 25 -> 50 -> 75 -> (HEAD)
cll.append(10)
cll.append(25)
cll.append(50)
cll.append(75)

print("Current Circular Linked List:")
cll.display_linkedlist()

# Case 1: Search for an element present at index 2 (value 50)
target = 50
idx = cll.find_index_of_node(target)
print(f"Index of element {target}: {idx}")  # Output: Index of element 50: 2

# Case 2: Search for the head element (value 10)
target = 10
idx = cll.find_index_of_node(target)
print(f"Index of element {target}: {idx}")  # Output: Index of element 10: 0

# Case 3: Search for an element not present in the list (value 99)
target = 99
idx = cll.find_index_of_node(target)
print(f"Index of element {target}: {idx}")  # Output: Index of element 99: -1
````

* **`current = self.head` & `curr_index = 0**`
**Cursor & Counter Setup:** Points the traversal pointer `current` to the head node address and sets the position counter `curr_index` to $0$.
* **`if current.data == target_value:` & `return curr_index**`
**Target Found Condition:** Dereferences `current.data` and compares it with `target_value`. If equal, the method immediately exits and returns the integer stored in `curr_index`.

**Pointer Dereferencing & Increment:** Advances `current` to the next memory node address and increments the 0-based index tracker by $1$.

**Full-Circle Guard:** Detects when `current` has wrapped around back to `self.head`. Breaking out of the loop prevents infinite circular traversal when searching for non-existent values.
* **`return -1`**
**Sentinel Return:** Returns `-1` to indicate that all $n$ nodes were evaluated without finding `target_value`.

> **Time Complexity:** $O(n)$ — Performs a linear scan over up to $n$ nodes in dynamic memory.
> **Space Complexity:** $O(1)$ — Uses a constant amount of extra memory (`current` pointer and `curr_index` integer).

## DELETE LINKEDLIST

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        """Appends a new node to the end of the circular linked list."""
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            new_node.next = self.head
            return

        current = self.head
        while current.next != self.head:
            current = current.next

        current.next = new_node
        new_node.next = self.head

    def delete_linkedlist(self):
        """Deletes all nodes in the circular linked list and clears dynamic memory."""
        if self.head is None:
            print("List is already empty!")
            return

        current = self.head

        # Step 1: Break the circular reference cycle
        # Locate tail node and sever its link back to self.head
        while current.next != self.head:
            current = current.next
        current.next = None  # Breaks the cycle into a linear sequence

        # Step 2: Unlink all individual node references sequentially
        current = self.head
        while current:
            next_node = current.next
            current.next = None  # Remove reference to next node
            current = next_node

        # Step 3: Clear the entrance pointer
        self.head = None
        print("Circular Linked List deleted successfully!")

    def display_linkedlist(self):
        """Prints the circular linked list contents."""
        if self.head is None:
            print("List is empty!")
            return

        current = self.head
        while True:
            print(f"{current.data} -> ", end="")
            current = current.next
            if current == self.head:
                break
        print("(HEAD)")


# --- Testing the implementation ---
cll = CircularLinkedList()

# Populate list: 10 -> 20 -> 30 -> 40 -> (HEAD)
cll.append(10)
cll.append(20)
cll.append(30)
cll.append(40)

print("List before deletion:")
cll.display_linkedlist()  # Output: 10 -> 20 -> 30 -> 40 -> (HEAD)

# Deleting the entire list
cll.delete_linkedlist()

print("\nList after deletion:")
cll.display_linkedlist()  # Output: List is empty!
````
